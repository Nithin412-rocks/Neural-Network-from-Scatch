```python
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
```

# Loading the Dataset and converting to tensors

```python
# we will load the dataset and convert it in to numpy array and than to tensor so that we can pass it to the neural network
df = pd.read_csv("/content/pima-indians-diabetes.csv", header = None, index_col=None)
df.head()
```

```python
X = torch.tensor(df.iloc[:,:8].to_numpy(), dtype = torch.float32)
Y = torch.tensor(df.iloc[:,8].to_numpy(), dtype = torch.float32).reshape(-1,1)
```

# Creating Neural Network from Torch

```python
# first define the layers
# second define the loss function
# thrid define the optimizer

class DemoNeuralNetwork(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden1 = nn.Linear(8, 12)
    self.act1 = nn.ReLU()
    self.hidden2 = nn.Linear(12, 8)
    self.act2 = nn.ReLU()
    self.output = nn.Linear(in_features=8, out_features=1)
    self.act3 = nn.Sigmoid()

  def forward(self,x):
    x = self.act1(self.hidden1(x))
    x = self.act2(self.hidden2(x))
    x = self.act3(self.output(x))
    return x

model = DemoNeuralNetwork()
print(model)
```

```python
# now lets define the loss and optimizer
loss_fn = nn.BCELoss()
optimizer = torch.optim.Adam(params=model.parameters(), lr=0.001)
```

```python
# Now lets define the training like epochs and batch size
# epoch is iteration through the entire dataset
# batch size is for that size gradient descent algorithm updates the weights

n_epochs = 100
batch_size = 10

for epoch in range(n_epochs):
  for i in range(X.shape[0]//batch_size):
    start = i * batch_size
    end = start + batch_size
    X_batch = X[start:end]
    Y_pred = model(X_batch)
    Y_batch = Y[start:end]
    loss = loss_fn(Y_pred, Y_batch)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
  print(f'Finished epoch {epoch}, latest loss {loss}')
```
